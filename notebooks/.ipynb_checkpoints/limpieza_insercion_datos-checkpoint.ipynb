```python
from sqlalchemy import create_engine, text
import pandas as pd
import unicodedata
import math


# funciones de limpieza de datos 
def limpiar_texto(texto, es_correo=False):
    if pd.isnull(texto):
        return None
    if es_correo: 
        return texto  #verificar si es correo y devolver tal cual el dato
    texto = str(texto).upper() # mayusculas
    texto = texto.replace("Ñ", "N").replace("_", "").replace("-", "") # remplazar caracteres especiales en los demas campos 
    texto = ''.join(
        c for c in unicodedata.normalize('NFD', texto) # Normalización de Forma Descompuesta, separa los acentos de las vocales ejemplo á ->  a+´
        if unicodedata.category(c) != 'Mn' # filtra las tildes (signos diacriticos)
    )
    return texto
    
def redondear_superior(valor):
    if pd.isnull(valor):
        return None
    return math.ceil(valor * 100) / 100 # regresar el numero redondeado y a su inmediato superior

# Parametros de conexion
db_host = 'localhost'
db_port = '5432'
db_name = 'mp3_database'
db_user = 'postgres'
db_pass = 'postgres'

# Crear el engine (conexion sql)
engine = create_engine(f'postgresql+psycopg2://{db_user}:{db_pass}@{db_host}:{db_port}/{db_name}')


# lectura de tablas desde postgressql
clientes_df = pd.read_sql("SELECT * FROM clientes", engine)
productos_df = pd.read_sql("SELECT * FROM productos", engine)
corresponsales_df = pd.read_sql("SELECT * FROM corresponsales", engine)
ordenes_df = pd.read_sql("SELECT * FROM ordenes", engine)
orden_detalles_df = pd.read_sql("SELECT * FROM orden_detalles", engine)

# limpiar los datos extraidos
clientes_df['nombre'] = clientes_df['nombre'].apply(limpiar_texto)
clientes_df['correo'] = clientes_df['correo'].apply(limpiar_texto, es_correo=True)  # el correo no se debe modificar
productos_df['nombre'] = productos_df['nombre'].apply(limpiar_texto)
productos_df['descripcion'] = productos_df['descripcion'].apply(limpiar_texto)
corresponsales_df['nombre'] = corresponsales_df['nombre'].apply(limpiar_texto)

# Redondear los valores numericos
productos_df['precio'] = productos_df['precio'].apply(redondear_superior)
ordenes_df['total_orden'] = ordenes_df['total_orden'].apply(redondear_superior)


# Creacion de tablas para modelo estrella
create_dim_producto_sql = """
CREATE TABLE IF NOT EXISTS dim_producto (
    id_producto INT PRIMARY KEY,
    nombre VARCHAR(255),
    descripcion TEXT
);
"""

create_dim_cliente_sql = """
CREATE TABLE IF NOT EXISTS dim_cliente (
    id_cliente INT PRIMARY KEY,
    nombre VARCHAR(255),
    correo VARCHAR(255),
    identificacion VARCHAR(13)
);
"""

create_dim_corresponsal_sql = """
CREATE TABLE IF NOT EXISTS dim_corresponsal (
    id_corresponsal INT PRIMARY KEY,
    nombre VARCHAR(255)
);
"""

create_dim_tiempo_sql = """
CREATE TABLE IF NOT EXISTS dim_tiempo (
    id_fecha DATE PRIMARY KEY,
    dia INT,
    mes INT,
    anio INT,
    trimestre INT
);
"""

create_fact_ventas_sql = """
CREATE TABLE IF NOT EXISTS hechos_ventas (
    id_venta SERIAL PRIMARY KEY,
    id_producto INT,
    id_cliente INT,
    id_corresponsal INT,
    id_fecha DATE,
    cantidad INT,
    total DECIMAL(10, 2),
    FOREIGN KEY (id_producto) REFERENCES dim_producto(id_producto),
    FOREIGN KEY (id_cliente) REFERENCES dim_cliente(id_cliente),
    FOREIGN KEY (id_corresponsal) REFERENCES dim_corresponsal(id_corresponsal),
    FOREIGN KEY (id_fecha) REFERENCES dim_tiempo(id_fecha)
);
"""

# Ejecutar la creación de las tablas
with engine.connect() as conn:
    conn.execute(text(create_dim_producto_sql))
    conn.execute(text(create_dim_cliente_sql))
    conn.execute(text(create_dim_corresponsal_sql))
    conn.execute(text(create_dim_tiempo_sql))
    conn.execute(text(create_fact_ventas_sql))
    conn.commit()

print("Tablas creadas correctamente.")


```

```python
#insertar los datos que se recolectaron anteriormente desde postgressql
def insertar_datos():
    with engine.connect() as conn:
        # insertar dim producto
        for _, row in productos_df.iterrows(): # iteramos las filas del df cliente
            conn.execute(
                text("INSERT INTO dim_producto (id_producto, nombre, descripcion) VALUES (:id_producto, :nombre, :descripcion)"),
                {"id_producto": row['id_producto'], "nombre": row['nombre'], "descripcion": row['descripcion']}
            )
        
        # insertar dim cliente
        for _, row in clientes_df.iterrows(): # iteramos las filas del df cliente
            conn.execute(
                text("INSERT INTO dim_cliente (id_cliente, nombre, correo, identificacion) VALUES (:id_cliente, :nombre, :correo, :identificacion)"),
                {"id_cliente": row['id_cliente'], "nombre": row['nombre'], "correo": row['correo'], "identificacion": row['identificacion']}
            )
        
        # insertar dim corresponsal
        for _, row in corresponsales_df.iterrows(): # iteramos las filas del df corresponsales
            conn.execute(
                text("INSERT INTO dim_corresponsal (id_corresponsal, nombre) VALUES (:id_corresponsal, :nombre)"),
                {"id_corresponsal": row['id_corresponsal'], "nombre": row['nombre']}
            )
        
        # insertar dim tiempo desde la fecha transaccion de ordenes
        ordenes_df['fecha_transaccion'] = pd.to_datetime(ordenes_df['fecha_transaccion'])
        for _, row in ordenes_df.iterrows(): # iteramos las filas del df ordenes
            conn.execute(
                text("INSERT INTO dim_tiempo (id_fecha, dia, mes, anio, trimestre) VALUES (:id_fecha, :dia, :mes, :anio, :trimestre)"),
                {"id_fecha": row['fecha_transaccion'].date(), # fecha completa
                 "dia": row['fecha_transaccion'].day, # dia
                 "mes": row['fecha_transaccion'].month, # mes
                 "anio": row['fecha_transaccion'].year, # anio
                 "trimestre": (row['fecha_transaccion'].month - 1) // 3 + 1} # trimestre
            )
        
        # insertar hechos ventas
        for _, row in ordenes_df.iterrows(): # iteramos las filas del df ordenes
            detalle = orden_detalles_df[orden_detalles_df['id_orden'] == row['id_orden']]
            for _, detalle_row in detalle.iterrows():  # iteramos las filas del df detalles orden
                conn.execute(
                    text("INSERT INTO hechos_ventas (id_producto, id_cliente, id_corresponsal, id_fecha, cantidad, total) VALUES (:id_producto, :id_cliente, :id_corresponsal, :id_fecha, :cantidad, :total)"),
                    {
                        "id_producto": int(detalle_row['id_producto']),
                        "id_cliente": int(row['id_cliente']),
                        "id_corresponsal": int(row['id_corresponsal']),
                        "id_fecha": row['fecha_transaccion'].date(),
                        "cantidad": int(detalle_row['cantidad']),        
                        "total": float(row['total_orden'])  # Asegúrate de que esté redondeado si lo deseas
    }
)
        
        conn.commit()
        print("Datos insertados correctamente")

# se llama la funcion para insertar los datos
insertar_datos()
```

```python
# pruebas tomando en cuenta lo solicitado
# producto mas vendido 
df_producto_mas_vendido = pd.read_sql_query("""SELECT 
    p.nombre AS producto,
    SUM(hv.cantidad) AS total_vendido
FROM hechos_ventas hv
JOIN dim_producto p ON hv.id_producto = p.id_producto
GROUP BY p.nombre
ORDER BY total_vendido DESC;
""", engine)
df_producto_mas_vendido.head()


```

```python
# clientes con el mayor numero de pedidos
df_producto_mas_vendido = pd.read_sql_query("""SELECT 
    c.nombre AS cliente,
    COUNT(DISTINCT CAST(hv.id_fecha AS TEXT) || CAST(hv.id_producto AS TEXT)) AS num_pedidos
FROM hechos_ventas hv
JOIN dim_cliente c ON hv.id_cliente = c.id_cliente
GROUP BY c.nombre
ORDER BY num_pedidos DESC;
""", engine)

df_producto_mas_vendido.head()
```
